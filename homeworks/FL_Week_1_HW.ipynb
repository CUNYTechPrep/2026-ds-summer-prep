# Q1: Alphabet Slices
* Store the first ten letters of the alphabet in a list.
* Use a slice to print out the first three letters of the alphabet.
* Use a slice to print out any three letters from the middle of your list.

```python
# Solve Alphabet slices here. 
## Extra Credit:  Do this without 'hard coding' the alpahbet.

import random 

firstTen = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j'] #first ten letters of alphabet
print(firstTen[0:3]) #Print the first three letters of the alphabet

#Print any three letters from the middle of list, but may return duplicates...!!!
middleList = []
i = 0
while i < 3: #loop runs 3x, to get three letters
    ra = random.choice(firstTen[1:-1]) #use import random, slice the list to exlcude the first and last letters
    middleList.append(ra) #append the randomly selected letter in the new list
    i += 1

print(middleList)
```

# Q2:  Covert all the rapper names to title case and save them into a new different list. 
Example: **lil wayne** becomes **Lil Wayne**

```python
# Solve rapper names here
rappers = ['lil wayne', 'nicki minaj', 'drake']

titleCase_rappers = [] #new list to store capitalized names
for name in rappers: #go through each name in the rapper list
    cap_name = name.title() #use title function to turn the first letter of each word into upper case
    titleCase_rappers.append(cap_name) #append the upper case word to the new list
print(titleCase_rappers)
```

# Q5: 
## Write a function that takes a list and returns a new list that has all the duplicates removed.

Example input and expected output:
- input = `["Michele", "Robin", "Sara", "Michele"]`
- expected output = `['Michele', 'Robin', 'Sara']`


```python
# Solve problem here:

names = ["Michele", "Robin", "Sara", "Michele"]

def remove_dup(names):
    new_name = [] #new list with the removed duplicates

    name_set = set(names) #turn the org list into a set - doesn't allow duplicates

    for name in name_set: #append each ele in the set to the new list
        new_name.append(name)
    return new_name #return the new list

#Test the function
names2 = remove_dup(names)
print(names2)
```

# Q6: Write a function that takes a list of numbers 
(for example, `a = [5, 10, 15, 20, 25]`) and returns a new list of only the first and last elements of the given list.

Example input and expected output:
- input = `[5, 10, 15, 20, 25]`
- expected output = `[5, 25]`

```python
# Solve problem here:
input_list = [5, 10, 99, 20, 25]

def first_last(input_list):
    if len(input_list) == 0: #account for if empty list, return empty list
        return []
    if len(input_list) == 1: #account for if only one ele in list
        return None
    else: #list has >= 2 elements
        new_list = []
        new_list.append(input_list[0]) #append first ele
        new_list.append(input_list[-1]) #append last ele
        return new_list


#Test the function
lst1 = first_last(input_list)
print(lst1)
```

# Q8: Write a function that takes a number as input and returns the following:
* If the input is divisible by three, return `'fizz'`
* If the input is divisible by five, return `'buzz'`
* If the input is divisible by three and by five, return `'fizzbuzz'`
* If the input is not divisible by three or five, return `None`.

```python
# Solve Problem fizzbuzz here:
def fizzbuzz(num):
    if num % 3 == 0 and num % 5 == 0: #need to check this condition first! - otherwise if check divisible by 3 or 5 only, will miss this check
        return "fizzbuzz"
    elif num % 5 == 0:
        return "buzz"
    elif num % 3 == 0:
        return "fizz"
    else:
        return None

print(fizzbuzz(3)) #fizz
print(fizzbuzz(25)) #buzz
print(fizzbuzz(30)) #fizzbuzz
print(fizzbuzz(2)) #None
```
