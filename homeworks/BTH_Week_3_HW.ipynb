### Exercise 5C — Return more than one result

Complete `score_summary(scores)` so it returns a tuple containing:

1. the lowest score
2. the highest score
3. the average score

Round the average to 2 decimal places.

```python
scores = [72, 88, 91, 79]
def score_summary(scores):
    # TODO
    lowest = min(scores)
    highest = max(scores)
    average = round(sum(scores) / len(scores), 2)
    return lowest, highest, average
```

## 7. Integrated pair challenge

**Switch Driver and Navigator halfway through this challenge.**

The data below contains student records. Complete `analyze_students(records)` so it returns a dictionary with exactly these keys:

- `"student_count"`: number of student records
- `"average_score"`: average of all scores, rounded to 2 decimal places
- `"passing_names"`: names of students with scores of 70 or higher, in their original order
- `"unique_skills"`: a set of all skills represented
- `"top_student"`: the name of the student with the highest score

Your function should work with other lists that use the same structure—not only this exact dataset.

Break the problem into small steps. You may create helper functions if they make your solution clearer.

```python
students = [
    {"name": "Aisha", "score": 88, "skills": ["Python", "SQL"]},
    {"name": "Luis", "score": 67, "skills": ["Python", "Git"]},
    {"name": "Maya", "score": 94, "skills": ["SQL", "Pandas"]},
    {"name": "Noah", "score": 76, "skills": ["Git", "Pandas"]},
]

def analyze_students(records):
    # TODO
    student_count = len(records)
    total = 0
    passing_names = []
    unique_skills = set()
    top_student = records[0]["name"]
    top_score = records[0]["score"]

    for student in records:
        total = total + student["score"]

        if student["score"] >= 70:
            passing_names.append(student["name"])

        for skill in student["skills"]:
            unique_skills.add(skill)

        if student["score"] > top_score:
            top_score = student["score"]
            top_student = student["name"]

    average_score = round(total / student_count, 2)

    return {
        "student_count": student_count,
        "average_score": average_score,
        "passing_names": passing_names,
        "unique_skills": unique_skills,
        "top_student": top_student,
    }

analysis = analyze_students(students)
analysis
```

```python
# Automated check — do not edit.
assert isinstance(analysis, dict)
assert set(analysis.keys()) == {
    "student_count", "average_score", "passing_names", "unique_skills", "top_student"
}
assert analysis["student_count"] == 4
assert analysis["average_score"] == 81.25
assert analysis["passing_names"] == ["Aisha", "Maya", "Noah"]
assert analysis["unique_skills"] == {"Python", "SQL", "Git", "Pandas"}
assert analysis["top_student"] == "Maya"

# A second dataset checks whether the function is general rather than hard-coded.
second_group = [
    {"name": "Devin", "score": 70, "skills": ["Python"]},
    {"name": "Sam", "score": 80, "skills": ["Python", "Git"]},
]
second_analysis = analyze_students(second_group)
assert second_analysis["student_count"] == 2
assert second_analysis["average_score"] == 75.0
assert second_analysis["passing_names"] == ["Devin", "Sam"]
assert second_analysis["unique_skills"] == {"Python", "Git"}
assert second_analysis["top_student"] == "Sam"
passed("7")
```

Complete `safe_average(values)` so it:

- returns the average for a non-empty list of numbers
- returns `None` for an empty list
- returns `None` rather than crashing if a value cannot be treated as a number

You may use `try` / `except`.

```python
def safe_average(values):
    # TODO
    if len(values) == 0:
        return None
    try:
        return sum(values) / len(values)
    except TypeError:
        return None

assert safe_average([10, 20, 30]) == 20
assert safe_average([]) is None
assert safe_average([10, "oops", 30]) is None
passed("Stretch B")
```
