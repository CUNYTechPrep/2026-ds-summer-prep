## #1 Figure out how to print the current date and time. 
The solution should be in this format:  2021-08-05 11:56:30 (YYYY-MM-DD HH:MM:SS)

Hint:  You're going to have to import a library

````python
from datetime import datetime
from zoneinfo import ZoneInfo
curr_time = datetime.now(ZoneInfo("US/Eastern")).strftime("%Y-%m-%d %H:%M:%S") #use US/Eastern timezone, use strftime to not show microsec
print(curr_time)
````

## #2 Write a function that accepts the radius of a circle and returns the area of a circle.  
The function for the area of a circle is pi * radius squared.  

````python
from math import pi
def find_area(rad):
    return pi * pow(rad, 2) #use power function

area = find_area(2)
print(area)
````

## #3 Write a function that takes a list of numbers and returns True of all number in the list are unique and returns False if the list contains duplicate numbers. 
* Hint, maybe think of using a set()

Input:  input_list = [2,4,5,5,7,9]

Expected Output:  False

````python
def is_unique(input_list):
    set_list = set(input_list) #turn the list into a set
    if len(input_list) == len(set_list): #if the len of the list and the set match, all the nums are unique
        return True
    else:
        return False #otherwise if the len are not equal, there are duplicate numbers

input_list = [2, 4, 5, 7, 7, 9, 9]
result = is_unique(input_list)
print(result)
````

## #4 Write a Python program to count the number of each character of a given text of a text file.
* Use the text file called 'lil-wayne.txt' in the data folder.  
* Convert all the characters to uppercase first, then count them.  
* Use a dictionary in which the keys of the dictionary are the characters, and the values are the counts of that character. 

Expected Result:  `{'D': 74, 'W': 63, 'A': 211, 'Y': 44, 'N': 165, 'E': 274, ' ': 522, 'M': 77, 'I': 185, 'C': 60, 'H': 122, 'L': 126, 'R': 152, 'T': 180, 'J': 8, '.': 26, '(': 12, 'B': 59, 'O': 145, 'S': 152, 'P': 39, '2': 24, '7': 7, ',': 37, '1': 28, '9': 16, '8': 6, ')': 12, '[': 14, ']': 14, 'K': 13, 'F': 44, 'G': 53, 'X': 5, 'U': 66, 'V': 20, '3': 2, '4': 4, '5': 5, '6': 3, '0': 36, '\n': 6, '-': 6, ';': 1, '!': 1, '"': 8, "'": 3, '—': 1}`

````python
from pathlib import Path

file_path = Path("../Week-02/intermediate/data/lil-wayne.txt") #get the correct path for the txt file

d = {} #create a dictionary

with open(file_path, "r") as file: #open the txt file
    while True:
        char = file.read(1) #read one char at a time
        if not char:
            break
        up_char = char.upper() #convert each char into uppercase
        if up_char not in d: #count each char in the text file
            d[up_char] = 1
        else:
            d[up_char] += 1

print(d) #print the dictionary

````

## #5 Using the list below...
* Remove the fourth number from the list.
* Take that number you just removed and multiply the last number of the list by that number. 
* Append that number to the end of the list.
* Print the list

INPUT: [1, 1, 2, 3, 5, 8, 13, 21]

EXPECTED OUTPUT:  [1, 1, 2, 5, 8, 13, 21, 63]



````python
list1 = [1, 1, 2, 3, 5, 8, 13, 21]

del_num = list1.pop(3) #remove the fourth element (index 3) and store it in a var
del_num *= list1[-1] #multiply the deleted num by the last ele (index -1)
list1.append(del_num) #append the multiplied deleted num to the end of the list
print(list1)
````

## #6 Using the list below, make a new list that contains only the numbers from the original list that are divisible by three.

* Input:  original_list = [2, 3, 6, 8, 9, 15, 19, 21]
* Expected new list:  [3, 6, 9, 15, 21]


````python
original_list = [2, 3, 6, 8, 9, 15, 19, 21]

div_by_three = [] #create a new list
for num in original_list: #go through each num in list, if it's divisible by 3, append it to new list
    if num % 3 == 0:
        div_by_three.append(num)

print(div_by_three)
````

## #9  Given a two Python list. Iterate both lists simultaneously such that list1 should display item in original order and list2 in reverse order.
Given input:
```
list1 = [10, 20, 30, 40]
list2 = [100, 200, 300, 400]
```

Expected Output:
```
10 400
20 300
30 200
40 100
```

````python
list1 = [10, 20, 30, 40]
list2 = [100, 200, 300, 400]

#use while loop
i = 0  #index for list1 ele
j = len(list2) - 1  #index for list2 ele
while i < len(list1) and j >= 0: #use < for list1, >= for list2, loop thru forward for l1 and backwards for l2
    print(list1[i], list2[j]) #print the ele on one line
    i += 1 #increment i 
    j -= 1 #decrement j
````
