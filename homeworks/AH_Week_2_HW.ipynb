## #3 Write a function that takes a list of numbers and returns True of all number in the list are unique and returns False if the list contains duplicate numbers. 
* Hint, maybe think of using a set()

Input:  input_list = [2,4,5,5,7,9]

Expected Output:  False

````python
def all_numbers_unique(numbers):
    unique_numbers = set(numbers)

    if len(numbers) == len(unique_numbers):
        return True
    else:
        return False


input_list = [2, 4, 5, 5, 7, 9]

print(all_numbers_unique(input_list))
````

## #4 Write a Python program to count the number of each character of a given text of a text file.
* Use the text file called 'lil-wayne.txt' in the data folder.  
* Convert all the characters to uppercase first, then count them.  
* Use a dictionary in which the keys of the dictionary are the characters, and the values are the counts of that character. 

Expected Result:  `{'D': 74, 'W': 63, 'A': 211, 'Y': 44, 'N': 165, 'E': 274, ' ': 522, 'M': 77, 'I': 185, 'C': 60, 'H': 122, 'L': 126, 'R': 152, 'T': 180, 'J': 8, '.': 26, '(': 12, 'B': 59, 'O': 145, 'S': 152, 'P': 39, '2': 24, '7': 7, ',': 37, '1': 28, '9': 16, '8': 6, ')': 12, '[': 14, ']': 14, 'K': 13, 'F': 44, 'G': 53, 'X': 5, 'U': 66, 'V': 20, '3': 2, '4': 4, '5': 5, '6': 3, '0': 36, '\n': 6, '-': 6, ';': 1, '!': 1, '"': 8, "'": 3, '—': 1}`

````python

file_path = "../Week-02/intermediate/data/lil-wayne.txt"

with open(file_path, "r", encoding="utf-8") as file:
    text = file.read()

text = text.upper()

character_counts = {}

for character in text:
    if character in character_counts:
        character_counts[character] += 1
    else:
        character_counts[character] = 1

print(character_counts)
````

# 4.5 Figure out how to sort the dictionary to find which character appears the most.

````python
sorted_characters = sorted(
    character_counts.items(),
    key=lambda item: item[1],
    reverse=True
)

print(sorted_characters)
print("Most common character:", sorted_characters[0])
````

## #6 Using the list below, make a new list that contains only the numbers from the original list that are divisible by three.

* Input:  original_list = [2, 3, 6, 8, 9, 15, 19, 21]
* Expected new list:  [3, 6, 9, 15, 21]


````python
original_list = [2, 3, 6, 8, 9, 15, 19, 21]

divisible_by_three = []

for number in original_list:
    if number % 3 == 0:
        divisible_by_three.append(number)

print(divisible_by_three)
````

## #9  Given a two Python list. Iterate both lists simultaneously such that list1 should display item in original order and list2 in reverse order.
Given input:
```
list1 = [10, 20, 30, 40]
list2 = [100, 200, 300, 400]
```

Expected Output:
```
10 400
20 300
30 200
40 100
```

````python
list1 = [10, 20, 30, 40]
list2 = [100, 200, 300, 400]

reversed_list2 = list2[::-1]

for first_number, second_number in zip(list1, reversed_list2):
    print(first_number, second_number)
````

## #10 Remove empty strings from the list of strings.
Input:  `list_of_strings = ['I', 'love', '', '', 'data', '', 'science', '!']`

Expected Output:  `['I', 'love', 'data', 'science', '!']`

````python
list_of_strings = ["I", "love", "", "", "data", "", "science", "!"]

non_empty_strings = []

for word in list_of_strings:
    if word != "":
        non_empty_strings.append(word)

print(non_empty_strings)
````
