## #3 Write a function that takes a list of numbers and returns True of all number in the list are unique and returns False if the list contains duplicate numbers. 
* Hint, maybe think of using a set()

Input:  input_list = [2,4,5,5,7,9]

Expected Output:  False

````python
def unique_numbers(l):
    #empty set = set()
    unique_set = set()
    for num in l:
        if num in unique_set:
            return False
        else:
            unique_set.add(num)

    return True

# .split() - separates the string where there is a space
# map(int,...) - applies int to every element returned by split()
# list() - converts map object into list
li = list(map(int,input("enter numbers separated by a space: ").split()))
print("your list: " , li)
print(unique_numbers(li))
````

## #4 Write a Python program to count the number of each character of a given text of a text file.
* Use the text file called 'lil-wayne.txt' in the data folder.  
* Convert all the characters to uppercase first, then count them.  
* Use a dictionary in which the keys of the dictionary are the characters, and the values are the counts of that character. 

Expected Result:  `{'D': 74, 'W': 63, 'A': 211, 'Y': 44, 'N': 165, 'E': 274, ' ': 522, 'M': 77, 'I': 185, 'C': 60, 'H': 122, 'L': 126, 'R': 152, 'T': 180, 'J': 8, '.': 26, '(': 12, 'B': 59, 'O': 145, 'S': 152, 'P': 39, '2': 24, '7': 7, ',': 37, '1': 28, '9': 16, '8': 6, ')': 12, '[': 14, ']': 14, 'K': 13, 'F': 44, 'G': 53, 'X': 5, 'U': 66, 'V': 20, '3': 2, '4': 4, '5': 5, '6': 3, '0': 36, '\n': 6, '-': 6, ';': 1, '!': 1, '"': 8, "'": 3, '—': 1}`

````python
from pathlib import Path

# Go up to the root folder, then down to the target file
file_path = Path("..") / "Week-02" / "beginner" / "data" / "lil-wayne.txt"

# put the file content into a variable
with open(file_path, "r", encoding="utf-8") as file:
    text_data = file.read()

    upper_text = text_data.upper()
    counts = {}
    #if the char is not a space, it will add it 
for char in upper_text:
    if char.isalnum():
        #if letter exists in counts, it grabs it, if not defaults to 0
        counts[char] = counts.get(char, 0) + 1

print(counts)


    


    


````

# 4.5 Figure out how to sort the dictionary to find which character appears the most.

````python
#counts.items() converts each pair into paired tuple *sorry zack*
#key=lambda item: item[1] is telling the computer to sort based on the count not the letter
#reverse = true tells the computer to sort in decesnding order
#sorted() processes the list
#dict() puts the output into a dictinary cleanly 
highest_frequency = dict(sorted(counts.items(), key=lambda item: item[1], reverse=True))
print(highest_frequency) 
````

## #6 Using the list below, make a new list that contains only the numbers from the original list that are divisible by three.

* Input:  original_list = [2, 3, 6, 8, 9, 15, 19, 21]
* Expected new list:  [3, 6, 9, 15, 21]


````python
original_list = [2, 3, 6, 8, 9, 15, 19, 21]
new_list = []

for num in original_list:
    if num % 3 == 0:
        new_list.append(num)

print(new_list)
````
