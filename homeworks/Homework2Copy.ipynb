## #2 Write a function that accepts the radius of a circle and returns the area of a circle.  
The function for the area of a circle is pi * radius squared.  

````python
def calculate_area(radius):
    if isinstance(radius, int) == False or isinstance(radius, float) == False:
        raise ValueError("ValueError: The given value is not able to be multiplied. Enter an integer or float as a valid radius.")
    else:
        return 3.1415926535 * radius * radius
````

## #4 Write a Python program to count the number of each character of a given text of a text file.
* Use the text file called 'lil-wayne.txt' in the data folder.  
* Convert all the characters to uppercase first, then count them.  
* Use a dictionary in which the keys of the dictionary are the characters, and the values are the counts of that character. 

Expected Result:  `{'D': 74, 'W': 63, 'A': 211, 'Y': 44, 'N': 165, 'E': 274, ' ': 522, 'M': 77, 'I': 185, 'C': 60, 'H': 122, 'L': 126, 'R': 152, 'T': 180, 'J': 8, '.': 26, '(': 12, 'B': 59, 'O': 145, 'S': 152, 'P': 39, '2': 24, '7': 7, ',': 37, '1': 28, '9': 16, '8': 6, ')': 12, '[': 14, ']': 14, 'K': 13, 'F': 44, 'G': 53, 'X': 5, 'U': 66, 'V': 20, '3': 2, '4': 4, '5': 5, '6': 3, '0': 36, '\n': 6, '-': 6, ';': 1, '!': 1, '"': 8, "'": 3, '—': 1}`

````python
from pathlib import Path

def get_char_count():
    file = Path("2026-ds-summer-prep/Week-02/intermediate/data/lil-wayne.txt")
    char_dict = {}
    with open(file, '+r') as text_file:
        for line in text_file:
            new_line = text_file.readline(line)
            new_line.upper()
            for char in new_line:
                if char in char_dict:
                    char_dict[char] = 1
                else:
                    char_dict[char] += 1
    return char_dict

````

# 4.5 Figure out how to sort the dictionary to find which character appears the most.

````python
def sort_dict(dict):
    if dict is None:
        raise ValueError("There isn't an existing input dictionary, please enter a valid dictionary object.")

    sorted_list = []
    for i in dict:
        sorted_list.append(i)
    return sorted(sorted_list)
````

## #7 Reverse the list below

````python
original_list = [2, 3, 6, 8, 9, 15, 19, 21]
new_list = []
counter = -1
while counter > 0-len(original_list)-1:
    new_list.append(original_list[counter])
    counter-=1

print(new_list)
````

Input:  `list_of_strings = ['I', 'love', '', '', 'data', '', 'science', '!']`
