# Question 1:  Write a function to find the class average of all the student scores. 

```python
def class_average(students):
    total = sum(student["score"] for student in students)
    return total / len(students)

students = [
    {"name": "Aisha", "score": 70},
    {"name": "Luis", "score": 80},
    {"name": "Maya", "score": 90},
]

print(class_average(students))
```

# Another Scope Problem
## Related trap: in-place methods often return `None`

Predict the values of `scores` and `sorted_scores`.

```python
scores = [88, 72, 95]
sorted_scores = scores.sort()

print("scores:", scores)
print("sorted_scores:", sorted_scores)
```

`.sort()` changes the existing list and returns `None`. If you want a separate sorted list, use `sorted(scores)`.

## Gotcha 7 — Conditionals that look right but are always true

Predict the result when `answer` is `"no"`.

```python
answer = "no"

if answer in ["yes", "y"]:
    response = "Continue"
else:
    response = "Stop"

print(response)
```

Python reads the condition as `(answer == "yes") or ("y")`. A non-empty string such as `"y"` is truthy, so the condition always succeeds.

### Repair it

Use membership testing so `"yes"` and `"y"` are the only accepted answers.
