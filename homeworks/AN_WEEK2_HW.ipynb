## #3 Write a function that takes a list of numbers and returns True of all number in the list are unique and returns False if the list contains duplicate numbers. 
* Hint, maybe think of using a set()

Input:  input_list = [2,4,5,5,7,9]

Expected Output:  False

````python
def is_unique(lst):
    seen = set()
    for num in lst:
        if num in seen:
            return False
        seen.add(num)
    return True

print(is_unique([2, 4, 5, 5, 7, 9]))  # False
````

## #4 Write a Python program to count the number of each character of a given text of a text file.
* Use the text file called 'lil-wayne.txt' in the data folder.  
* Convert all the characters to uppercase first, then count them.  
* Use a dictionary in which the keys of the dictionary are the characters, and the values are the counts of that character. 

Expected Result:  `{'D': 74, 'W': 63, 'A': 211, 'Y': 44, 'N': 165, 'E': 274, ' ': 522, 'M': 77, 'I': 185, 'C': 60, 'H': 122, 'L': 126, 'R': 152, 'T': 180, 'J': 8, '.': 26, '(': 12, 'B': 59, 'O': 145, 'S': 152, 'P': 39, '2': 24, '7': 7, ',': 37, '1': 28, '9': 16, '8': 6, ')': 12, '[': 14, ']': 14, 'K': 13, 'F': 44, 'G': 53, 'X': 5, 'U': 66, 'V': 20, '3': 2, '4': 4, '5': 5, '6': 3, '0': 36, '\n': 6, '-': 6, ';': 1, '!': 1, '"': 8, "'": 3, '—': 1}`

````python

def count_letters():
    with open("../Week-02/beginner/data/lil-wayne.txt", "r", encoding="utf-8") as file:
        text = file.read()
    letter_count = {}
    for char in text:
        letter = char.upper()   
        letter_count[letter] = letter_count.get(letter, 0) + 1
    return letter_count
print (count_letters())
````

# 4.5 Figure out how to sort the dictionary to find which character appears the most.

````python
def most_common_letter():
    letter_count = count_letters()
    most_common = max(letter_count, key=letter_count.get)
    return most_common, letter_count[most_common]

print (most_common_letter())
````

## #5 Using the list below...
* Remove the fourth number from the list.
* Take that number you just removed and multiply the last number of the list by that number. 
* Append that number to the end of the list.
* Print the list

INPUT: [1, 1, 2, 3, 5, 8, 13, 21]

EXPECTED OUTPUT:  [1, 1, 2, 5, 8, 13, 21, 63]



````python

def update_list(lst):
    num = lst[3]
    lst.remove(num)
    lst.append(num * lst[len(lst) - 1])
    return lst

list1 = [1, 1, 2, 3, 5, 8, 13, 21]
print(update_list(list1))  # [1, 1, 2, 5, 8, 13, 21, 63]
````

## #6 Using the list below, make a new list that contains only the numbers from the original list that are divisible by three.

* Input:  original_list = [2, 3, 6, 8, 9, 15, 19, 21]
* Expected new list:  [3, 6, 9, 15, 21]


````python
def divisible_by_3(lst):
    divisible_by_3 = [num for num in lst if num % 3 == 0]
    return divisible_by_3

original_list = [2, 3, 6, 8, 9, 15, 19, 21]
print(divisible_by_3(original_list))  # [3, 6, 9, 15, 21]
````

## #8 Using the list below, turn every item of the list into its square.  
* Input original_list = [1, 2, 3, 4, 5, 6, 7]
* Expected new_list:  [1, 4, 9, 16, 25, 36, 49]


````python
def list_of_squares(nums):
    return [num ** 2 for num in nums]

original_list = [1, 2, 3, 4, 5, 6, 7]
print(list_of_squares(original_list))  # [1, 4, 9, 16, 25, 36, 49]
````

## #9  Given a two Python list. Iterate both lists simultaneously such that list1 should display item in original order and list2 in reverse order.
Given input:
```
list1 = [10, 20, 30, 40]
list2 = [100, 200, 300, 400]
```

Expected Output:
```
10 400
20 300
30 200
40 100
```

````python
def merge_lists(lst1, lst2):
    while lst1 and lst2:
        print(lst1.pop(0), lst2.pop())
    return

list1 = [10, 20, 30, 40]
list2 = [100, 200, 300, 400]
merge_lists(list1, list2)  # [10, 400, 20, 300, 30, 300, 40, 100]
````

## #10 Remove empty strings from the list of strings.
Input:  `list_of_strings = ['I', 'love', '', '', 'data', '', 'science', '!']`

Expected Output:  `['I', 'love', 'data', 'science', '!']`

````python
def remove_empty_strings(lst):
    for string in lst:
        if string == '':
            lst.remove(string)
    return lst
list_of_strings = ['I', 'love', '', '', 'data', '', 'science', '!']
print(remove_empty_strings(list_of_strings))  # ['I', 'love', 'data', 'science', '!']
````

## What was the oldest and youngest ages on the titanic?

````python
def oldest_youngest(df):
    oldest = df['age'].max()
    youngest = df['age'].min()
    return float(oldest), float(youngest)
print (oldest_youngest(df))
````
