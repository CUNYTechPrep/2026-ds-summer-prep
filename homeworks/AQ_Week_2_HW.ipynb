## #1 Figure out how to print the current date and time. 
The solution should be in this format:  2021-08-05 11:56:30 (YYYY-MM-DD HH:MM:SS)

## #2 Write a function that accepts the radius of a circle and returns the area of a circle.  
The function for the area of a circle is pi * radius squared.  

````python
import math

def circle_area(radius):
    """Calculate the area of a circle given its radius."""
    if radius < 0:
        raise ValueError("Radius cannot be negative.")
    return math.pi * (radius ** 2)

print("Area of circle with radius 5:", circle_area(5))
print("Area of circle with radius 0:", circle_area(0))
print("Area of circle with radius 10:", circle_area(10))
````

## #3 Write a function that takes a list of numbers and returns True of all number in the list are unique and returns False if the list contains duplicate numbers. 
* Hint, maybe think of using a set()

Input:  input_list = [2,4,5,5,7,9]

Expected Output:  False

````python
def check_unique_numbers(numbers):
    """Check if all numbers in the list are unique."""
    return len(numbers) == len(set(numbers))

print("Are [1, 2, 3, 4] unique?", check_unique_numbers([1, 2, 3, 4]))
print("Are [1, 2, 2, 4] unique?", check_unique_numbers([1, 2, 2, 4]))
````

## #4 Write a Python program to count the number of each character of a given text of a text file.
* Use the text file called 'lil-wayne.txt' in the data folder.  
* Convert all the characters to uppercase first, then count them.  
* Use a dictionary in which the keys of the dictionary are the characters, and the values are the counts of that character. 

Expected Result:  `{'D': 74, 'W': 63, 'A': 211, 'Y': 44, 'N': 165, 'E': 274, ' ': 522, 'M': 77, 'I': 185, 'C': 60, 'H': 122, 'L': 126, 'R': 152, 'T': 180, 'J': 8, '.': 26, '(': 12, 'B': 59, 'O': 145, 'S': 152, 'P': 39, '2': 24, '7': 7, ',': 37, '1': 28, '9': 16, '8': 6, ')': 12, '[': 14, ']': 14, 'K': 13, 'F': 44, 'G': 53, 'X': 5, 'U': 66, 'V': 20, '3': 2, '4': 4, '5': 5, '6': 3, '0': 36, '\n': 6, '-': 6, ';': 1, '!': 1, '"': 8, "'": 3, '—': 1}`

````python
from pathlib import Path

matches = list(Path.cwd().rglob("lil-wayne.txt"))
if not matches:
    raise FileNotFoundError("Could not find 'lil-wayne.txt' anywhere under the current workspace.")

file_path = matches[0]

with file_path.open("r", encoding="utf-8") as file:
    content = file.read().upper()

dictionary = {}
for char in content:
    dictionary[char] = dictionary.get(char, 0) + 1

print(dictionary)
````

## #5 Using the list below...
* Remove the fourth number from the list.
* Take that number you just removed and multiply the last number of the list by that number. 
* Append that number to the end of the list.
* Print the list

INPUT: [1, 1, 2, 3, 5, 8, 13, 21]

EXPECTED OUTPUT:  [1, 1, 2, 5, 8, 13, 21, 63]



````python
list1 = [1, 1, 2, 3, 5, 8, 13, 21]
print("INPUT: {0}".format(list1))

fourth_element = None
for i in range(len(list1)):
    if i == 3:
        fourth_element = list1[i]
        list1.remove(list1[i])
        break

product = list1[len(list1) - 1] * fourth_element

list1.append(product)
print("EXPECTED OUTPUT: {0}".format(list1))
````

## #6 Using the list below, make a new list that contains only the numbers from the original list that are divisible by three.

* Input:  original_list = [2, 3, 6, 8, 9, 15, 19, 21]
* Expected new list:  [3, 6, 9, 15, 21]


````python
original_list = [2, 3, 6, 8, 9, 15, 19, 21]
new_list = []
print("Input: {0}".format(original_list))

for i in range(len(original_list)):
    if original_list[i] % 3 == 0:
        new_list.append(original_list[i])

print("Expected new list: {0}".format(new_list))
````

## #9  Given a two Python list. Iterate both lists simultaneously such that list1 should display item in original order and list2 in reverse order.

```
list1 = [10, 20, 30, 40]
list2 = [100, 200, 300, 400]
```

Expected Output:
```
10 400
20 300
30 200
40 100
```
