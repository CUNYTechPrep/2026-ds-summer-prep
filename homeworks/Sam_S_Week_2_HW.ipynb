## #3 Write a function that takes a list of numbers and returns True of all number in the list are unique and returns False if the list contains duplicate numbers. 
* Hint, maybe think of using a set()

Input:  input_list = [2,4,5,5,7,9]

Expected Output:  False

````python
def dupes(nums : int)-> bool:
    seen = set()

    for num in nums:
        if num in seen:
            return False
        
        seen.add(num)

    return True

# Another cool way to do it with constant space complexity

def dupes_v2(nums: int)-> bool:
    nums.sort() # .sort() sorts in place so we don't use extra space

    for num1, num2 in zip(nums, nums[1:]):
        if num1 == num2:
            return False

    return True


test_case1 = [0,1,2,3,4,5,6,7,8]

test_case2 = [0,1,2,2,2,2]

print(dupes(test_case1))

print(dupes(test_case2))

print("==== Second set of test cases ====")

print(dupes_v2(test_case1))
print(dupes_v2(test_case2))
````

## #9  Given a two Python list. Iterate both lists simultaneously such that list1 should display item in original order and list2 in reverse order.
Given input:
```
list1 = [10, 20, 30, 40]
list2 = [100, 200, 300, 400]
```

Expected Output:
```
10 400
20 300
30 200
40 100
```

````python
list1 = [10, 20, 30, 40]
list2 = [100, 200, 300, 400]

for l1, l2 in zip(list1, list2[::-1]):
    print(f"{l1}  {l2}")

print("=== Approach 2 ===")

l1 = 0
l2 = len(list2) -1

while l1 <= len(list1) and l2 >= 0:
    l1_word = list1[l1]
    l2_word = list2[l2]

    print(f"{l1_word}  {l2_word}")

    l1 += 1
    l2 -= 1
````

## #10 Remove empty strings from the list of strings.
Input:  `list_of_strings = ['I', 'love', '', '', 'data', '', 'science', '!']`

Expected Output:  `['I', 'love', 'data', 'science', '!']`

````python
list_of_strings = ['I', 'love', '', '', 'data', '', 'science', '!']

non_empty_strings = []

for string in list_of_strings:
    if string:
        non_empty_strings.append(string)

print(non_empty_strings)
````
