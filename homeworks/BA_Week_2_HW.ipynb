## #3 Write a function that takes a list of numbers and returns True of all number in the list are unique and returns False if the list contains duplicate numbers. 
* Hint, maybe think of using a set()

Input:  input_list = [2,4,5,5,7,9]

Expected Output:  False

````python
test = [2,4,5,5,7,9]

def is_unique_list(lst):
    tmpSet = set()
    for item in lst:
        if item in tmpSet:
            return False
        tmpSet.add(item)
    return True

is_unique_list(test)
````

## #4 Write a Python program to count the number of each character of a given text of a text file.
* Use the text file called 'lil-wayne.txt' in the data folder.  
* Convert all the characters to uppercase first, then count them.  
* Use a dictionary in which the keys of the dictionary are the characters, and the values are the counts of that character. 

Expected Result:  `{'D': 74, 'W': 63, 'A': 211, 'Y': 44, 'N': 165, 'E': 274, ' ': 522, 'M': 77, 'I': 185, 'C': 60, 'H': 122, 'L': 126, 'R': 152, 'T': 180, 'J': 8, '.': 26, '(': 12, 'B': 59, 'O': 145, 'S': 152, 'P': 39, '2': 24, '7': 7, ',': 37, '1': 28, '9': 16, '8': 6, ')': 12, '[': 14, ']': 14, 'K': 13, 'F': 44, 'G': 53, 'X': 5, 'U': 66, 'V': 20, '3': 2, '4': 4, '5': 5, '6': 3, '0': 36, '\n': 6, '-': 6, ';': 1, '!': 1, '"': 8, "'": 3, '—': 1}`

````python

file_path = '/workspaces/2026-ds-summer-prep/Week-02/intermediate/data/lil-wayne.txt'
fh = open(file_path, 'r')

tmp_freq_dict = {}
for line in fh:
    for char in line:
        cap_char = char.upper()
        if cap_char not in tmp_freq_dict:
            tmp_freq_dict[cap_char] = 1
        else:
            tmp_freq_dict[cap_char] += 1

tmp_freq_dict
````

# 4.5 Figure out how to sort the dictionary to find which character appears the most.

````python
# tmp_freq_dict.items() returns a set of lists ordered by key, value
# sorted() func returns a list sorted by a given key (if given)
# the key in this case is based on a given list within the previous set from 
#                                                       tmp_freq_dict.items()
# this key is given a variable named pair_element, and we specify the second 
#               index as we want to sort by the int value attached to the key
#  


sorted_tmp_dict = dict(
                    sorted(
                        tmp_freq_dict.items(), 
                        key=lambda pair_element: pair_element[1], 
                        reverse=True
                    )
                )
sorted_tmp_dict
````

## #7 Reverse the list below

````python
original_list = [2, 3, 6, 8, 9, 15, 19, 21]

for index, item in enumerate(original_list):
    if index == (len(original_list) // 2):
        break
    rear_item = original_list[-1+(-1*index)]
    original_list[index] = rear_item
    original_list[-1+(-1*index)] = item

original_list
````
