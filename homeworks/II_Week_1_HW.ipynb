# Q2:  Covert all the rapper names to title case and save them into a new different list. 
Example: **lil wayne** becomes **Lil Wayne**

```python
# Solve rapper names here
rappers = ['lil wayne', 'nicki minaj', 'drake']

#john payne pork
title_case_rappers = []

for full_name in rappers:
    # splits the full_name into separate names
    split_name = full_name.split()

    # will go through the split names and change their first letter to uppercase.
    for i in range(len(split_name)):
        name = split_name[i]
        split_name[i] = name[0:1].upper() + name[1::]
        
    # recombimes the split name and appends it to the result list
    title_case_rappers.append(" ".join(split_name))

print(title_case_rappers)

            

```

# Q4: Find the sum and the average of this list of numbers.

Try doing this using a loop.  Then try doing this without using a loop. 

```python
# Solve problem here:

my_list = [1, 5, 10, 55, 88, 44, 42, 50, 20, 38]
list_sum = 0
list_avg = 0

# With for loop
for num in my_list:
    # adds numbers
    list_sum += num 

#average
list_avg = list_sum / len(my_list)

# Without loop

list_sum = sum(my_list)
list_avg = list_sum / len(my_list)

# Keep this as your last line in this cell.
print(list_sum, list_avg)
```

# Q8: Write a function that takes a number as input and returns the following:
* If the input is divisible by three, return `'fizz'`
* If the input is divisible by five, return `'buzz'`
* If the input is divisible by three and by five, return `'fizzbuzz'`
* If the input is not divisible by three or five, return `None`.

```python
# Solve Problem fizzbuzz here:

def fizzbuzz(num: int) -> str:

    if num % 3 == 0:
        if num % 5 == 0:
            return "fizzbuzz"
        #returns if divisible by just 3
        return "fizz"
    # returns if divisible by just 5
    return "buzz"

print(fizzbuzz(15))
```
