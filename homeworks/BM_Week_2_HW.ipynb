````python
# Shows working directory
import os
print(os.getcwd())
````

````python
def character_count(filename):
  with open(filename, 'r') as file: #opens and reads file
    text = file.read()
    text = text.upper()
  
  count = {}
  for char in text:
    if char in count:
      count[char] = count[char] + 1 # adds to current count of characters
    else:
      count[char] = 1 # 1st apperance of letter
  return count

lil_wayne = character_count('/Users/bryanmogrovejo/2026-ds-summer-prep/Week-02/intermediate/data/lil-wayne.txt')
print(lil_wayne)
````

## #6 Using the list below, make a new list that contains only the numbers from the original list that are divisible by three.

* Input:  original_list = [2, 3, 6, 8, 9, 15, 19, 21]
* Expected new list:  [3, 6, 9, 15, 21]

## #9  Given a two Python list. Iterate both lists simultaneously such that list1 should display item in original order and list2 in reverse order.
Given input:
```
list1 = [10, 20, 30, 40]
list2 = [100, 200, 300, 400]
```

Expected Output:
```
10 400
20 300
30 200
40 100
```

````python
list1 = [10, 20, 30, 40]
list2 = [100, 200, 300, 400]

for num,num2 in zip(list1, reversed(list2)): #zip() allows multiple lists to be looped through while pairing at same position
  print(num, num2)
````

## #10 Remove empty strings from the list of strings.
Input:  `list_of_strings = ['I', 'love', '', '', 'data', '', 'science', '!']`

Expected Output:  `['I', 'love', 'data', 'science', '!']`

````python
list_of_strings = ['I', 'love', '', '', 'data', '', 'science', '!']
clean_list = []
for word in list_of_strings:
  if word != '':
    clean_list.append(word)

print(clean_list)
````
