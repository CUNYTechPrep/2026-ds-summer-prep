1. the lowest score

3. the average score

Round the average to 2 decimal places.

```python
scores = [72, 88, 91, 79]
def score_summary(scores):
    # TODO
    min_score = min(scores)
    max_score = max(scores)
    avg_score = sum(scores)/len(scores)
    score_tuple = (min_score, max_score, avg_score) #put the min, max, and avg scores in the tuple
    return score_tuple
```

Your function should work with other lists that use the same structure—not only this exact dataset.

```python
students = [
    {"name": "Aisha", "score": 88, "skills": ["Python", "SQL"]},
    {"name": "Luis", "score": 67, "skills": ["Python", "Git"]},
    {"name": "Maya", "score": 94, "skills": ["SQL", "Pandas"]},
    {"name": "Noah", "score": 76, "skills": ["Git", "Pandas"]},
]

#Helper functions

#Num of student records
def count_students(students):
    stu_count = 0
    for student in students: #count each dict ele within the list
        stu_count += 1
    return stu_count

#avg of all scores, rounded to 2 dec places
def find_avg(students):
    total_score = 0
    avg_score = 0
    for i, student in enumerate(students): #go through the list - each dict ele, enumerate the list
        total_score += students[i]["score"] #use the STUDENTS list (NOT student), accumulate each score into total, use each ele (dict) score - access with index
    avg_score = total_score / len(students)
    rounded_avg = round(avg_score, 2)
    return rounded_avg

#names of students with scores >= 70 in their ORG order
def pass_names(students):
    list1 = [] #lists are ordered!!!
    for i, student in enumerate(students):
        if students[i]["score"] >= 70: #score needs to be >= 70
            list1.append(students[i]["name"]) #append the stu name
    return list1


#a set of all skills - no dups
def list_skills(students):
    skill_set = set() #create a new set
    for i, student in enumerate(students):
        for skill in students[i]["skills"]:
            if skill not in skill_set:
                skill_set.add(skill) #add the skill in the set if it's not already included
    return skill_set

#name of student with the highest score
def find_top(students):
#Part 1: save the name:score as a dictionary
    d = {}
    for i, student in enumerate(students):
        if students[i]["name"] not in d:
            d[students[i]["name"]] = students[i]["score"]
        else:
            d[students[i]["name"]] = students[i]["score"]

#Part 2: find the name of the student with the highest score
    for key, value in d.items():
        max_score = max(d, key = d.get) #find the key that has the highest value using max and get
    return max_score


def analyze_students(records):
    stu_results = {} #new dictionary to hold results

    #Call helper function to find num of student records
    s_count = count_students(records) #store result in s_count 
    stu_results["student_count"] = s_count
   

    #Call helper function to find the average of all scores
    s_avg = find_avg(records)
    stu_results["average_score"] = s_avg
  

    #Call helper function to find the names of students who scores >= 70 (ORG order)
    s_pass = pass_names(records)
    stu_results["passing_names"] = s_pass
   

    #Call helper function to find the unique skills (set)
    s_skills = list_skills(records)
    stu_results["unique_skills"] = s_skills
   

    #Call helper function to find the student name with the highest score
    s_top = find_top(records)
    stu_results["top_student"] = s_top


    return stu_results #return the dictionary with the results
    

analysis = analyze_students(students) #pass students list, in functs use records as the parameter name
analysis
```
