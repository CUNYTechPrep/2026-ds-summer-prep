# Q1: Alphabet Slices
* Store the first ten letters of the alphabet in a list.
* Use a slice to print out the first three letters of the alphabet.
* Use a slice to print out any three letters from the middle of your list.

```python
# Solve Alphabet slices here. 
## Extra Credit:  Do this without 'hard coding' the alpahbet.

alphabet = [chr(ord('a') + i) for i in range(10)]
print(alphabet[:3])
print(alphabet[3:6])
```

# Q4: Find the sum and the average of this list of numbers.

Try doing this using a loop.  Then try doing this without using a loop. 

```python
# Solve problem here:

my_list = [1, 5, 10, 55, 88, 44, 42, 50, 20, 38]
list_sum = 0
list_avg = 0

# this is another way to do it.
list_sum_new = sum(my_list)
list_avg_new = sum(my_list) / len(my_list)
print(list_sum_new, list_avg_new)

# # this is the loop way to do it.
temp = 0

for i in my_list: 
    temp += i
    list_sum = temp

list_avg = list_sum / len(my_list)

# Keep this as your last line in this cell.
print(list_sum, list_avg)
```

# Q6: Write a function that takes a list of numbers 
(for example, `a = [5, 10, 15, 20, 25]`) and returns a new list of only the first and last elements of the given list.

Example input and expected output:
- input = `[5, 10, 15, 20, 25]`
- expected output = `[5, 25]`

```python
# Solve problem here:
input_list = [5, 10, 99, 20, 25]

def first_last(lst):
    return [lst[0], lst[-1]]

print(first_last(input_list))
```
