## #3 Write a function that takes a list of numbers and returns True of all number in the list are unique and returns False if the list contains duplicate numbers. 
* Hint, maybe think of using a set()

Input:  input_list = [2,4,5,5,7,9]

Expected Output:  False

````python
def unique(x):
    b = list(set(x))
    x.sort()
    if b == x:
        return True
    else:
        return False

x = [2,4,5,3,6,67,5,7,9]

unique(x)

````

## #4 Write a Python program to count the number of each character of a given text of a text file.
* Use the text file called 'lil-wayne.txt' in the data folder.  
* Convert all the characters to uppercase first, then count them.  
* Use a dictionary in which the keys of the dictionary are the characters, and the values are the counts of that character. 

Expected Result:  `{'D': 74, 'W': 63, 'A': 211, 'Y': 44, 'N': 165, 'E': 274, ' ': 522, 'M': 77, 'I': 185, 'C': 60, 'H': 122, 'L': 126, 'R': 152, 'T': 180, 'J': 8, '.': 26, '(': 12, 'B': 59, 'O': 145, 'S': 152, 'P': 39, '2': 24, '7': 7, ',': 37, '1': 28, '9': 16, '8': 6, ')': 12, '[': 14, ']': 14, 'K': 13, 'F': 44, 'G': 53, 'X': 5, 'U': 66, 'V': 20, '3': 2, '4': 4, '5': 5, '6': 3, '0': 36, '\n': 6, '-': 6, ';': 1, '!': 1, '"': 8, "'": 3, '—': 1}`

````python
with open("../Week-02/intermediate/data/lil-wayne.txt", "r") as file:    #open file with read permission and set it as file
    content = file.read()                 #stores the file into content

with open("../Week-02/intermediate/data/lil-wayne.txt", "w") as file:           #open file with write permission
    file.write(content.upper())                  #overwrites file with .upper()  <-- overwrites all into uppercase into that file

dictionary = {}

with open("../Week-02/intermediate/data/lil-wayne.txt", "r") as file:
    for word in file:                 #each word in file
        for letter in word:                        #each letter in word
            if letter not in dictionary:            # if letter not in dictionary, #append it as a (letter): 1
                dictionary[letter] = 1             
            else:                                       #or else, append the existing letter by 1 
                dictionary[letter] += 1                 

print(dictionary)           #printing dictionary 
````

# 4.5 Figure out how to sort the dictionary to find which character appears the most.

````python
import numpy as np

k = list(dictionary.keys())                    #set keys and values as seperate list
d = list(dictionary.values())

idx = reversed(np.argsort(d))            #sort the array from least to most (the array that forms it is the index of it) and reverses the index

res = {k[i]: d[i] for i in idx}          #for loop and forms a new dictionary as you trace through the reversed index

print(res)
````

## #9  Given a two Python list. Iterate both lists simultaneously such that list1 should display item in original order and list2 in reverse order.
Given input:
```
list1 = [10, 20, 30, 40]
list2 = [100, 200, 300, 400]
```

Expected Output:
```
10 400
20 300
30 200
40 100
```

````python
list1 = [10, 20, 30, 40]
list2 = [100, 200, 300, 400]

list2.reverse()

count = 0
n = len(list1) - 1


while count <= n:
    print(f'{list1[count]} {list2[count]}')
    count += 1
````
