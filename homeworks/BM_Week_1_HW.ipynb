# Q1: Alphabet Slices
* Store the first ten letters of the alphabet in a list.
* Use a slice to print out the first three letters of the alphabet.
* Use a slice to print out any three letters from the middle of your list.

```python
# Solve Alphabet slices here. 
## Extra Credit:  Do this without 'hard coding' the alpahbet.

# imports the string library to be able to get the alphabet (thanks google)
import string
# stores entire lowercase alphabet into a list
ascii_letters = list(string.ascii_lowercase)
print("Alphabet: ", ascii_letters)

# stores the first 10 letters into a seperate list
first_ten = ascii_letters[0:10]
print("First 10 letters: ",first_ten)

# stores the first 3 letters into a seperate list
first_three = ascii_letters[0:3]
print("First 3 letters: ", first_three)

# stores the middle 3 letters from the "first_ten" list into a seperate list
middle_three = first_ten[3:6]
print("Middle 3 letters from the first 10 letters: ", middle_three)
```

# Q2:  Covert all the rapper names to title case and save them into a new different list. 
Example: **lil wayne** becomes **Lil Wayne**

```python
# Solve rapper names here
rappers = ['lil wayne', 'nicki minaj', 'drake']
print(rappers)

# copies the list of rappers and turns it into a string 
str_rappers = ", ".join(rappers)
print(str_rappers)

# titleize's the string 
str_rappers = str_rappers.title()
print(str_rappers)

# copies the string and turns it into a new list
list_rappers = str_rappers.split(", ")
print(list_rappers)
```

# Q3: Write a function that takes a number and returns:
* True if the input number is even.
* False if the input number is odd.

```python
# Solve problem here

def even_odd(num):
   '''Checks if a number is even and returns true or odd and returns false'''
   # if any number mod 2 is equal to 0, it's even
   if num % 2 == 0:
      return True
   # if not equal to 0, it must be odd
   else: 
      return False
even_odd(3)
```

# Q4: Find the sum and the average of this list of numbers.

Try doing this using a loop.  Then try doing this without using a loop. 

```python
# Solve problem here:

# without loop
my_list = [1, 5, 10, 55, 88, 44, 42, 50, 20, 38]
list_sum = sum(my_list)
list_avg = sum(my_list) / len(my_list)

print(list_sum, list_avg)

# with loop

# initialize 2 variables to 0 to keep track inside the loop
total = 0
count = 0
for i in my_list:
    total += i # sum of all numbers in my_list
    count += 1 # keeps track of how many numbers to divide the sum by and find the average
average = total / count
print(total)
print(average)


```

# Q5: 
## Write a function that takes a list and returns a new list that has all the duplicates removed.

Example input and expected output:
- input = `["Michele", "Robin", "Sara", "Michele"]`
- expected output = `['Michele', 'Robin', 'Sara']`


```python
# Solve problem here:

names = ["Michele", "Robin", "Sara", "Michele"]

def remove_duplicates(x):
    '''Removes any duplicates'''
    unique_names = set(x) # converts the parameter into a set since sets can't have dupes
    return list(unique_names) # converts the set back into a list and returns the list aka outputs it

remove_duplicates(names)
```

# Q6: Write a function that takes a list of numbers 
(for example, `a = [5, 10, 15, 20, 25]`) and returns a new list of only the first and last elements of the given list.

Example input and expected output:
- input = `[5, 10, 15, 20, 25]`
- expected output = `[5, 25]`

```python
# Solve problem here:
input_list = [5, 10, 99, 20, 25]

def first_last(x):
    '''Returns the first and last elements'''
    first = x[0] # 0 gives us first element 
    last = x[-1] # -1 gives us last element 

    return [first, last]

first_last(input_list)
```

# Q8: Write a function that takes a number as input and returns the following:
* If the input is divisible by three, return `'fizz'`
* If the input is divisible by five, return `'buzz'`
* If the input is divisible by three and by five, return `'fizzbuzz'`
* If the input is not divisible by three or five, return `None`.

```python
# Solve Problem fizzbuzz here:
def classic_leetcode(num):
    '''Fizzbuzz question'''
    if num % 3 == 0 and num % 5 == 0:
        return "fizzbuzz"
    elif num % 3 == 0:
        return "fizz"
    elif num % 5 == 0:
        return "buzz"
    else:
        return None

classic_leetcode(30)
```
