## #3 Write a function that takes a list of numbers and returns True of all number in the list are unique and returns False if the list contains duplicate numbers. 
* Hint, maybe think of using a set()

Input:  input_list = [2,4,5,5,7,9]

Expected Output:  False

````python
def has_duplicate(lst : list) -> bool:
    s = set()
    for item in lst:
        s.add(item)
    return True if (len(s) != len(lst)) else False

l1 = [1,2,3]
l2 = [1, 2, 3, 1]
print(f"does l1 have duplicates? {has_duplicate(l1)}")
print(f"Does l2 have duplicates? {has_duplicate(l2)}")
````

# 4.5 Figure out how to sort the dictionary to find which character appears the most.

````python

import re

# first time implementing radix sort.
# changed my mind, I should be sorting by the counts, not the words themselves.
def sortByCount(counts : dict[str, int]) -> tuple[dict[int, list], int]:
    sorted = {}
    max_count = 0
    for word, count in counts.items():
        if(count not in sorted):
            sorted[count] = [word]
        else:
            sorted[count].append(word)
        if(count > max_count):
            max_count = count
    return (sorted, max_count)


with open("/workspaces/2026-ds-summer-prep/Week-02/intermediate/data/lil-wayne.txt", 'r') as f:
    counts = {}
    for line in f:
        words = re.split(r"\s\s*", line.upper())
        for word in words:
            if(word in counts):
                counts[word] += 1
            else:
                counts[word] = 1


sorted, max_count = sortByCount(counts)
print(f"words which appeared the most in the text: {sorted[max_count]}")
````

##  #11 Check if the value 999 exists in the dicitionary below.  If so, print true, if not print false.

Input: sample_dict = {'a': 100, 'b': 999, 'c': 300}

Expected Output:  True

````python
sample_dict = {'a': 100, 'b': 999, 'c': 300}
false_dict = {'a': 100, 'b': 200, 'c': 300}

def check_999(d : dict[str, int]) -> bool:
    for key, value in d.items():
        if(value == 999):
            return True
    return False

check_999(sample_dict)
check_999(false_dict)
````

## What was the oldest and youngest ages on the titanic?

````python
import pandas as pd

# read_csv() returns a 'dataframe', w/e that is.
df = pd.read_csv('/workspaces/2026-ds-summer-prep/Week-02/intermediate/data/titanic.csv')
min_age = df['age'][0]
max_age = df['age'][0]
for age in df['age']:
    if(pd.notna(age)):
        if(age < min_age):
            min_age = age
        elif(age > max_age):
            max_age = age
print(f"min_age: {min_age}")
print(f"max_age: {max_age}")
````
