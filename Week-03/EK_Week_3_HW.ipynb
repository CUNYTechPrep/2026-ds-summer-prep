### Exercise 5C — Return more than one result

Complete `score_summary(scores)` so it returns a tuple containing:

1. the lowest score
2. the highest score
3. the average score

Round the average to 2 decimal places.

```python
scores = [72, 88, 91, 79]
def score_summary(scores):
    scores1 = (min(scores), max(scores), sum(scores) / len(scores))
    return scores1
```

**Pair checkpoint:** In your own words, what is local scope? Why can relying heavily on global variables make a program harder to understand?

> Your answer: A local scope is what in the program has access to certain variables. For example, if a variable is created inside of a function, it can only be accessed within that function. An error will be thrown is the variable does not exist globally, and if it does, it will be overridden when it is called outside the function. Relying on global variables can make a program harder to understand because it means that variable names cannot be reused or their values will be overridden by the most recent assignment. This is not an issue for local variables within functions, where you can have as many of them with the same name as you'd like as long as they are all in different functions.

```python
students = [
    {"name": "Aisha", "score": 88, "skills": ["Python", "SQL"]},
    {"name": "Luis", "score": 67, "skills": ["Python", "Git"]},
    {"name": "Maya", "score": 94, "skills": ["SQL", "Pandas"]},
    {"name": "Noah", "score": 76, "skills": ["Git", "Pandas"]},
]

def analyze_students(records):
    student_count = len(records)
    average_score = getAverageScore(records)
    passing_names = getPassingNames(records)
    unique_skills = getUniqueSkills(records)
    top_student = getHighestScore(records)
    
    return {
        "student_count": student_count,
        "average_score": average_score,
        "passing_names": passing_names,
        "unique_skills": unique_skills,
        "top_student": top_student
    }


def getAverageScore(records):
    total = 0
    for record in records:
        total += record["score"]
    return total / len(records)

def getHighestScore(records):
    highestScore = records[0]["score"]
    highest = records[0]["name"]
    for record in records:
        if(record["score"] > highestScore):
            highest = record["name"]
            highestScore = record["score"]
    return highest

def getPassingNames(records):
    passing = []
    for record in records:
        if(record["score"] > 69):
            passing.append(record["name"])
    return passing

def getUniqueSkills(records):
    uniqueSkills = []
    for record in records:
        for skill in record["skills"]:
            uniqueSkills.append(skill)

    return set(uniqueSkills)

analysis = analyze_students(students)
analysis
```

Complete `safe_average(values)` so it:

- returns the average for a non-empty list of numbers
- returns `None` for an empty list
- returns `None` rather than crashing if a value cannot be treated as a number

You may use `try` / `except`.

```python
def safe_average(values):
    try:
        if(len(values) == 0):
            return None

        return sum(values) / len(values)
    except TypeError:
        return None

assert safe_average([10, 20, 30]) == 20
assert safe_average([]) is None
assert safe_average([10, "oops", 30]) is None
passed("Stretch B")
```
