## Gotcha — Local scope and `UnboundLocalError`

**Switch Driver and Navigator.**

This is a particularly confusing error: `students` exists outside the function, and you can use it inside the function, confusingly. 

Predict what happens, then run the cell.

```python
######### Write a function that takes in a list of dicts of student records
### Return a dict with: 
# 'student_count' as number of studednts, 
# 'average_score' as class average, 
# 'top_student' as name of top scorer
### 

students = [
    {"name": "Aisha", "score": 70},
    {"name": "Luis", "score": 80},
    {"name": "Maya", "score": 90},
]

def analyze_students(records):

    student_count = len(students)

    total_score = 0
    top_score = 0
    top_student = ""

    for row in students:
        total_score += row['score']
        if row["score"] >= top_score:
            top_score = row["score"]
            top_student = row['name']

    average_score = total_score / student_count

    solution_dict = {
        "average_score": average_score,
        "top_student": top_student,
        "student_count": student_count
    }
    return solution_dict


analysis = analyze_students(students)
print(analysis)
```

```python
second_group = [
    {"name": "Devin", "score": 70},
    {"name": "Sam", "score": 80},
]

analysis = analyze_students(second_group)
print(analysis)
```

## Related trap: in-place methods often return `None`

Predict the values of `scores` and `sorted_scores`.

```python
scores = [88, 72, 95]
sorted_scores = scores.sort()

print("scores:", scores)
print("sorted_scores:", sorted_scores)
```

`.sort()` changes the existing list and returns `None`. If you want a separate sorted list, use `sorted(scores)`.

## Gotcha 7 — Conditionals that look right but are always true

Predict the result when `answer` is `"no"`.

```python
answer = "no"

if answer == "yes" or "y":
    response = "Continue"
else:
    response = "Stop"

print(response)
```

Python reads the condition as `(answer == "yes") or ("y")`. A non-empty string such as `"y"` is truthy, so the condition always succeeds.

### Repair it

Use membership testing so `"yes"` and `"y"` are the only accepted answers.
